```python
RANDOM_SEED = 42
TRAIN_RATIO = 0.80

def split_dataset(raw_dataset, output_dataset):

    if raw_dataset is None:
        print("❌ Dataset is not available yet.")
        return

    image_dir = os.path.join(raw_dataset, "images")
    label_dir = os.path.join(raw_dataset, "labels")

    if not os.path.exists(image_dir):
        print("❌ images/ folder not found.")
        return

    if not os.path.exists(label_dir):
        print("❌ labels/ folder not found.")
        return

    image_extensions = ["*.jpg", "*.jpeg", "*.png", "*.JPG", "*.JPEG", "*.PNG"]

    image_files = []

    for ext in image_extensions:
        image_files.extend(glob.glob(os.path.join(image_dir, ext)))

    image_files = sorted(image_files)

    print("Total images found:", len(image_files))

    random.seed(RANDOM_SEED)
    random.shuffle(image_files)

    split_index = int(len(image_files) * TRAIN_RATIO)

    train_images = image_files[:split_index]
    val_images = image_files[split_index:]

    print("Training images:", len(train_images))
    print("Validation images:", len(val_images))
```

```python
# Create the final YOLO dataset with an 80/20 train-validation split

import os
import shutil
import random

RAW_IMAGES = "/content/NAYAN_Raw_Dataset/images/train"
RAW_LABELS = "/content/NAYAN_Raw_Dataset/labels/train"

FINAL_DATASET = "/content/NAYAN_Model_Training/dataset"

# Fixed seed for reproducibility
RANDOM_SEED = 42
TRAIN_RATIO = 0.80

# Create required directories
for folder in [
    os.path.join(FINAL_DATASET, "images", "train"),
    os.path.join(FINAL_DATASET, "images", "val"),
    os.path.join(FINAL_DATASET, "labels", "train"),
    os.path.join(FINAL_DATASET, "labels", "val")
]:
    os.makedirs(folder, exist_ok=True)

# Get all images
image_extensions = (".jpg", ".jpeg", ".png", ".JPG", ".JPEG", ".PNG")

image_files = [
    f for f in os.listdir(RAW_IMAGES)
    if f.endswith(image_extensions)
]

# Sort first so the same seed always produces the same split
image_files = sorted(image_files)

print("Total images:", len(image_files))

# Shuffle using fixed seed
random.seed(RANDOM_SEED)
random.shuffle(image_files)

# 80/20 split
split_index = int(len(image_files) * TRAIN_RATIO)

train_images = image_files[:split_index]
val_images = image_files[split_index:]

print("Training images:", len(train_images))
print("Validation images:", len(val_images))

# Copy image + matching label
missing_labels = []

for image_name in train_images:
    label_name = os.path.splitext(image_name)[0] + ".txt"

    source_image = os.path.join(RAW_IMAGES, image_name)
    source_label = os.path.join(RAW_LABELS, label_name)

    destination_image = os.path.join(
        FINAL_DATASET, "images", "train", image_name
    )
    destination_label = os.path.join(
        FINAL_DATASET, "labels", "train", label_name
    )

    shutil.copy2(source_image, destination_image)

    if os.path.exists(source_label):
        shutil.copy2(source_label, destination_label)
    else:
        missing_labels.append(label_name)


for image_name in val_images:
    label_name = os.path.splitext(image_name)[0] + ".txt"

    source_image = os.path.join(RAW_IMAGES, image_name)
    source_label = os.path.join(RAW_LABELS, label_name)

    destination_image = os.path.join(
        FINAL_DATASET, "images", "val", image_name
    )
    destination_label = os.path.join(
        FINAL_DATASET, "labels", "val", label_name
    )

    shutil.copy2(source_image, destination_image)

    if os.path.exists(source_label):
        shutil.copy2(source_label, destination_label)
    else:
        missing_labels.append(label_name)


# Final report
print("\n--- Split Summary ---")
print("Train images:", len(train_images))
print("Val images:", len(val_images))
print("Missing labels:", len(missing_labels))

if len(missing_labels) == 0:
    print("\n✅ 80/20 split completed successfully!")
    print("✅ Every image was paired with its label.")
else:
    print("\n⚠️ Missing labels detected:")
    for label in missing_labels[:20]:
        print(" ", label)
```

```python
import os
import tensorflow as tf

FP16_TFLITE = os.path.join(
    WORKSPACE,
    "yolov8n_nayan_50epochs",
    "weights",
    "best_fp16.tflite"
)

converter = tf.lite.TFLiteConverter.from_saved_model(SAVED_MODEL_DIR)

converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.target_spec.supported_types = [tf.float16]

tflite_fp16 = converter.convert()

with open(FP16_TFLITE, "wb") as f:
    f.write(tflite_fp16)

size_mb = os.path.getsize(FP16_TFLITE) / (1024 * 1024)

print("✅ FP16 TFLite model created!")
print("Path:", FP16_TFLITE)
print(f"Size: {size_mb:.2f} MB")
```

```python
import os
import shutil
from onnx2tf import convert

ONNX_PATH = "/content/NAYAN_Model_Training/yolov8n_nayan_50epochs/weights/best.onnx"

SAVED_MODEL_DIR = "/content/NAYAN_Model_Training/yolov8n_nayan_50epochs/weights/best_saved_model"

# Remove incomplete output if it exists
if os.path.exists(SAVED_MODEL_DIR):
    shutil.rmtree(SAVED_MODEL_DIR)

print("Starting ONNX → TensorFlow conversion...")
print("ONNX exists:", os.path.exists(ONNX_PATH))

try:
    convert(
        input_onnx_file_path=ONNX_PATH,
        output_folder_path=SAVED_MODEL_DIR
    )
    print("\nConversion command finished.")
except Exception as e:
    print("\n❌ Conversion failed:")
    print(type(e).__name__, ":", e)

print("\nChecking output...")
if os.path.exists(SAVED_MODEL_DIR):
    for root, dirs, files in os.walk(SAVED_MODEL_DIR):
        level = root.replace(SAVED_MODEL_DIR, "").count(os.sep)
        indent = "  " * level
        print(indent + os.path.basename(root) + "/")
        for file in files[:10]:
            print(indent + "  " + file)
else:
    print("❌ No SavedModel directory was created.")
```

```python
import os
import shutil
import csv

WEIGHTS_DIR = "/content/NAYAN_Model_Training/yolov8n_nayan_50epochs/weights"

BEST_PT = os.path.join(WEIGHTS_DIR, "best.pt")
FP32_TFLITE = os.path.join(WEIGHTS_DIR, "best.tflite")
INT8_TFLITE = os.path.join(WEIGHTS_DIR, "best_int8.tflite")

OUT_DIR = "/content/NAYAN_GitHub_Deliverables"

os.makedirs(OUT_DIR, exist_ok=True)

# Copy valid model files
shutil.copy2(
    BEST_PT,
    os.path.join(OUT_DIR, "nayan_yolov8n_best.pt")
)

shutil.copy2(
    FP32_TFLITE,
    os.path.join(OUT_DIR, "nayan_yolov8n_fp32.tflite")
)

shutil.copy2(
    INT8_TFLITE,
    os.path.join(OUT_DIR, "nayan_yolov8n_int8.tflite")
)

# --------------------------------------------------
# Comparison CSV
# --------------------------------------------------

comparison_file = os.path.join(
    OUT_DIR,
    "model_comparison.csv"
)

rows = [
    [
        "FP32 LiteRT",
        "nayan_yolov8n_fp32.tflite",
        11.70,
        0.3743,
        0.2367,
        "Validated"
    ],
    [
        "INT8 LiteRT",
        "nayan_yolov8n_int8.tflite",
        3.18,
        0.3592,
        0.1982,
        "Validated"
    ],
    [
        "FP16",
        "Not produced",
        "",
        "",
        "",
        "Separate FP16 TFLite export unsupported by current Ultralytics LiteRT exporter"
    ]
]

with open(comparison_file, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow([
        "Model",
        "File",
        "Size_MB",
        "mAP50",
        "mAP50_95",
        "Status"
    ])
    writer.writerows(rows)

# --------------------------------------------------
# README
# --------------------------------------------------

readme = """# NAYAN — Model Training & Quantization

## Training Configuration

- Model: YOLOv8n
- Dataset: 500 images
- Train/Validation split: 80/20
- Random seed: 42
- Training epochs: 50
- Image size: 640x640
- Batch size: 16
- Device: Google Colab GPU
- Number of classes: 11

## Classes

1. person
2. bicycle
3. car
4. motorcycle
5. bus
6. truck
7. traffic light
8. fire hydrant
9. stop sign
10. bench
11. chair

## Validation Results

| Model | Size | mAP50 | mAP50-95 |
|---|---:|---:|---:|
| Original YOLOv8n best.pt | — | 0.3779 | 0.2488 |
| FP32 LiteRT | 11.70 MB | 0.3743 | 0.2367 |
| INT8 LiteRT | 3.18 MB | 0.3592 | 0.1982 |

## Quantization Note

The current Ultralytics LiteRT exporter used in this experiment supports FP32 and INT8 export, but does not support creating a separate FP16 TFLite artifact through `quantize=16`.

An earlier ONNX-to-TFLite FP16 conversion produced an invalid detection output and was therefore excluded from the final deliverables rather than reporting an incorrect model.

The FP32 and INT8 LiteRT models included here were directly exported and validated.

## Files

- `nayan_yolov8n_best.pt`
- `nayan_yolov8n_fp32.tflite`
- `nayan_yolov8n_int8.tflite`
- `model_comparison.csv`

## Dataset

The dataset itself is not included in this repository because it is approximately 500 MB and is not required for the GitHub code contribution.

The dataset was split into:

- 400 training images
- 100 validation images

with corresponding YOLO label files.
"""

with open(
    os.path.join(OUT_DIR, "README.md"),
    "w"
) as f:
    f.write(readme)

print("✅ GitHub deliverables created:")
for name in sorted(os.listdir(OUT_DIR)):
    path = os.path.join(OUT_DIR, name)
    print(f"   {name} — {os.path.getsize(path)/(1024*1024):.2f} MB")
```
